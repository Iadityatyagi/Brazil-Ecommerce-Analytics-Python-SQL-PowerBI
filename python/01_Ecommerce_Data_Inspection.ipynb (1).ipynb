```python

# STEP 11: INVESTIGATE GEOLOCATION DUPLICATES

geo = pd.read_csv(
    data_path / "olist_geolocation_dataset.csv"
)

print("Total rows:", len(geo))

print(
    "Exact duplicate rows:",
    geo.duplicated().sum()
)

print("\nFirst 10 records:")
display(geo.head(10))

print("\nDuplicate ZIP code prefixes:")
print(
    geo["geolocation_zip_code_prefix"]
    .value_counts()
    .head(10)
)
```

```python

# STEP 12: FOREIGN KEY VALIDATION

def check_relationship(child_df, child_column,
                       parent_df, parent_column,
                       relationship_name):

    unmatched = (
        ~child_df[child_column].isin(parent_df[parent_column])
    ).sum()

    print("\nRelationship:", relationship_name)
    print("Unmatched records:", unmatched)


# Customer -> Orders
check_relationship(
    orders, "customer_id",
    customers, "customer_id",
    "Orders to Customers"
)

# Orders -> Order Items
check_relationship(
    order_items, "order_id",
    orders, "order_id",
    "Order Items to Orders"
)

# Orders -> Payments
check_relationship(
    payments, "order_id",
    orders, "order_id",
    "Payments to Orders"
)

# Orders -> Reviews
check_relationship(
    reviews, "order_id",
    orders, "order_id",
    "Reviews to Orders"
)

# Products -> Order Items
check_relationship(
    order_items, "product_id",
    products, "product_id",
    "Order Items to Products"
)

# Sellers -> Order Items
check_relationship(
    order_items, "seller_id",
    sellers, "seller_id",
    "Order Items to Sellers"
)
```

```python

# STEP 15: INVESTIGATE GEOLOCATION DUPLICATES

# 1. Check exact duplicate records
print("Exact duplicate rows:")
print(geo.duplicated().sum())

# 2. Check repeated ZIP code prefixes
print("\nTotal ZIP code records:")
print(len(geo))

print("\nUnique ZIP code prefixes:")
print(geo["geolocation_zip_code_prefix"].nunique())

# 3. Count repeated ZIP code prefixes
zip_counts = (
    geo["geolocation_zip_code_prefix"]
    .value_counts()
)

print("\nMost repeated ZIP code prefixes:")
print(zip_counts.head(10))

# 4. Create a summary by ZIP code
geo_summary = (
    geo.groupby("geolocation_zip_code_prefix")
    .agg(
        latitude=("geolocation_lat", "mean"),
        longitude=("geolocation_lng", "mean")
    )
    .reset_index()
)

print("\nGeolocation summary:")
print(geo_summary.head())

print("\nSummary rows:", len(geo_summary))
print(
    "Duplicate ZIP prefixes after grouping:",
    geo_summary["geolocation_zip_code_prefix"].duplicated().sum()
)
```

```python
# Convert purchase date if needed
orders_final["order_purchase_timestamp"] = pd.to_datetime(
    orders_final["order_purchase_timestamp"],
    errors="coerce"
)

# Add month to order items through order ID
monthly_data = items_final.merge(
    orders_final[["order_id", "order_purchase_timestamp"]],
    on="order_id",
    how="left"
)

monthly_sales = (
    monthly_data
    .dropna(subset=["order_purchase_timestamp"])
    .assign(
        month=lambda df: df["order_purchase_timestamp"].dt.to_period("M").astype(str)
    )
    .groupby("month")
    .agg(
        revenue=("price", "sum"),
        orders=("order_id", "nunique")
    )
    .reset_index()
)

display(monthly_sales.head(15))
```

```python
# STEP 32: CUSTOMER BEHAVIOUR ANALYSIS

customer_orders = orders_final.merge(
    customers_final[
        ["customer_id", "customer_unique_id",
         "customer_city", "customer_state"]
    ],
    on="customer_id",
    how="left"
)

print("Customer Orders Shape:", customer_orders.shape)

display(customer_orders.head())
```

```python
customer_frequency = (
    customer_orders
    .groupby("customer_unique_id")
    .agg(
        total_orders=("order_id", "nunique"),
        first_purchase=("order_purchase_timestamp", "min"),
        last_purchase=("order_purchase_timestamp", "max")
    )
    .reset_index()
)

print("Customer Frequency Shape:", customer_frequency.shape)

display(customer_frequency.head(10))
```

```python
repeat_customers = customer_frequency[
    customer_frequency["total_orders"] > 1
].copy()

one_time_customers = customer_frequency[
    customer_frequency["total_orders"] == 1
].copy()

print("Total Unique Customers:",
      len(customer_frequency))

print("Repeat Customers:",
      len(repeat_customers))

print("One-Time Customers:",
      len(one_time_customers))

print("Repeat Customer Percentage:",
      round(
          len(repeat_customers) / len(customer_frequency) * 100,
          2
      ), "%")
```

```python
# Attach customer identifiers to individual purchased items

customer_item_sales = items_final.merge(
    orders_final[["order_id", "customer_id"]],
    on="order_id",
    how="left"
)

customer_item_sales = customer_item_sales.merge(
    customers_final[
        ["customer_id", "customer_unique_id"]
    ],
    on="customer_id",
    how="left"
)

customer_revenue = (
    customer_item_sales
    .groupby("customer_unique_id")
    .agg(
        total_revenue=("price", "sum"),
        total_items=("order_item_id", "count"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

customer_revenue = customer_revenue.sort_values(
    "total_revenue",
    ascending=False
)

display(customer_revenue.head(10))
```

```python
customer_analysis = customer_frequency.merge(
    customer_revenue,
    on="customer_unique_id",
    how="left",
    suffixes=("_frequency", "_revenue")
)

customer_analysis["average_order_value"] = (
    customer_analysis["total_revenue"] /
    customer_analysis["total_orders_frequency"]
)

display(customer_analysis.head(10))
```

```python
# STEP 33: DELIVERY PERFORMANCE ANALYSIS

delivery_data = orders_final[
    [
        "order_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
].copy()

# Convert date columns
date_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    delivery_data[column] = pd.to_datetime(
        delivery_data[column],
        errors="coerce"
    )

print("Delivery Dataset Shape:", delivery_data.shape)

display(delivery_data.head(10))
```

```python
delivery_data["actual_delivery_days"] = (
    delivery_data["order_delivered_customer_date"] -
    delivery_data["order_purchase_timestamp"]
).dt.total_seconds() / 86400

display(
    delivery_data[
        ["order_id", "actual_delivery_days"]
    ].head(10)
)
```

```python
delivery_data["delivery_delay_days"] = (
    delivery_data["order_delivered_customer_date"] -
    delivery_data["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

display(
    delivery_data[
        [
            "order_id",
            "actual_delivery_days",
            "delivery_delay_days"
        ]
    ].head(10)
)
```

```python
delivery_data["delivery_performance"] = np.select(
    [
        delivery_data["order_status"].eq("delivered") &
        delivery_data["delivery_delay_days"].notna() &
        (delivery_data["delivery_delay_days"] > 0),

        delivery_data["order_status"].eq("delivered") &
        delivery_data["delivery_delay_days"].notna() &
        (delivery_data["delivery_delay_days"] <= 0)
    ],
    [
        "Late Delivery",
        "On-Time or Early"
    ],
    default="Not Evaluated"
)

print(
    delivery_data["delivery_performance"].value_counts()
)
```

```python
delivered_orders = delivery_data[
    delivery_data["order_status"] == "delivered"
].copy()

valid_delivery = delivered_orders.dropna(
    subset=[
        "actual_delivery_days",
        "delivery_delay_days"
    ]
)

total_delivered = len(delivered_orders)
total_evaluated = len(valid_delivery)

late_orders = (
    valid_delivery["delivery_delay_days"] > 0
).sum()

on_time_orders = (
    valid_delivery["delivery_delay_days"] <= 0
).sum()

average_delivery_days = (
    valid_delivery["actual_delivery_days"].mean()
)

late_delivery_rate = (
    late_orders / total_evaluated * 100
    if total_evaluated > 0 else np.nan
)

on_time_delivery_rate = (
    on_time_orders / total_evaluated * 100
    if total_evaluated > 0 else np.nan
)

print("========== DELIVERY KPIs ==========")
print("Total Delivered Orders:", total_delivered)
print("Orders Evaluated:", total_evaluated)
print("Late Orders:", late_orders)
print("On-Time or Early Orders:", on_time_orders)
print("Average Delivery Days:", round(average_delivery_days, 2))
print("Late Delivery Rate:", round(late_delivery_rate, 2), "%")
print("On-Time Delivery Rate:", round(on_time_delivery_rate, 2), "%")
```

```python
delivery_data.to_csv(
    cleaned_folder / "delivery_performance_analysis.csv",
    index=False
)

print(
    "Delivery Analysis Saved:",
    (cleaned_folder / "delivery_performance_analysis.csv").exists()
)
```

```python
review_distribution = (
    satisfaction_data
    .groupby("review_score")
    .agg(
        total_reviews=("review_id", "count")
    )
    .reset_index()
)

review_distribution["percentage"] = (
    review_distribution["total_reviews"] /
    review_distribution["total_reviews"].sum() * 100
).round(2)

display(review_distribution)
```

```python
review_category = (
    reviews_final[
        ["order_id", "review_score"]
    ]
    .merge(
        items_final[["order_id", "product_id"]],
        on="order_id",
        how="inner"
    )
    .merge(
        products_final[["product_id", "product_category_name"]],
        on="product_id",
        how="left"
    )
    .merge(
        category_final,
        on="product_category_name",
        how="left"
    )
)

review_category["product_category_name_english"] = (
    review_category["product_category_name_english"]
    .fillna("unknown")
)

display(review_category.head(10))
```
